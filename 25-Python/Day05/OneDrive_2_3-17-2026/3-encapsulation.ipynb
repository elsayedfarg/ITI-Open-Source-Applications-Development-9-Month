## Access Modifiers in Python

Unlike other languages like PHP, Python does not have explicit access modifiers
such as `public`, `protected`, and `private`. Instead, it uses naming conventions:

- **Public**: Attributes and methods with names starting with letters (`a-z`, `A-Z`) are public.
- **Protected**: Attributes prefixed with a single underscore (`_`) indicate that they should not be accessed directly (convention only).
- **Private**: Attributes prefixed with double underscores (`__`) are name-mangled to prevent accidental access.

Example of PHP-like class in Python:

```python
class Employee:
    """
    Employee class demonstrating encapsulation with access modifiers.
    """
    def __init__(self, name, email, salary):
        self.name = name       # Public attribute
        self._email = email    # Protected attribute (by convention, should not be accessed directly) class , childs
        self.__salary = salary  # Private attribute (name-mangled)

    def get_salary(self):
        """Getter method to retrieve the salary."""
        return self.__salary

    def get_email(self):
        """Getter method to retrieve the email."""
        return self._email

    def set_salary(self, sal):
        """Setter method to update the salary with validation."""
        if isinstance(sal, (int, float)) and sal > 10000:
            self.__salary = sal
        else:
            raise ValueError("Invalid salary. Salary must be a positive number.")

    #sat email
```

## Using the Property Decorator

Python provides a cleaner way to define getter and setter methods using the `@property` decorator.


```python
class EmployeeWithProperty:
    """
    Employee class demonstrating @property for cleaner getter and setter methods.
    """
    def __init__(self, name, email, salary):
        self.name = name
        self._email = email
        self.__salary = salary  # Uses the setter for validation

    @property
    def salary(self):
        """Getter method using @property. self.salary = salary"""
        return self.__salary

    @salary.setter
    def salary(self, sal):
        """Setter method with validation."""
        if isinstance(sal, (int, float)) and sal > 0:
            self.__salary = sal
        else:
            raise ValueError("Invalid salary. Salary must be a positive number.")

emp1 = Employee("", "", 0)
emp1.salary = 10000
print(emp1.salary)
```

```python
# Creating an instance
emp = EmployeeWithProperty("Mohamed", "moha@gmail.com", 100000)
print(emp.salary)  # Calls getter method
```

```python
emp.salary = 150000  # Calls setter method
print(emp.salary)
```

## Adding a Calculated Property: Age

We can add a calculated property `age` that depends on the employee's birth year.

```python
class EmployeeWithAge:
    """
    Employee class with dynamically calculated `age` property.
    """
    def __init__(self, name, email, salary, birth_year):
        self.name = name
        self._email = email
        self.__salary = salary
        self.birth_year = birth_year

    @property
    def age(self):
        """Calculates age dynamically based on the birth year."""
        return 2026 - self.birth_year

    @property
    def salary(self):
        return self.__salary

    @salary.setter
    def salary(self, sal):
        if isinstance(sal, (int, float)) and sal > 0:
            self.__salary = sal
        else:
            raise ValueError("Invalid salary. Salary must be a positive number.")
```

```python
# Creating an instance
emp = EmployeeWithAge("Mohamed", "moha@gmail.com", 90000, 1992)
print(emp.salary)  # Access salary
print(emp.age)  # Access calculated age
```
